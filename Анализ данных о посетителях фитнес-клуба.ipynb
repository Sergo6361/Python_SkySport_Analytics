```python
age_lst = [23, 45, 18, 80, 22, 25, 34, 37, 41, 88, 19, 20, 18, 45, 41, 39, 31, 32, 22, 28, 37]
```

```python
# Рассчитываем средний возраст по всему списку age_lst
avg_age = sum (age_lst) / len (age_lst)   # Суммируем все значения списка и делим их на длину списка
print (f"Средний возраст всех пользователей {round (avg_age, 2)} лет")   # Выводим получившийся результат
```

```python
# Рассчитываем средний возраст для посетителей, которые моложе 50 лет
sum_age_50 = 0 # Вводим новые переменные
cnt_age_50 = 0
for i in age_lst: # Запускаем цикл для фильтрации условия
    if i < 50:
        sum_age_50 += i
        cnt_age_50 += 1
avg_age_50 = sum_age_50 / cnt_age_50 # Суммируем все значения списка и делим их на длину списка
print (f"Средний возраст пользователей моложе 50 лет {round (avg_age_50, 2)}") # Выводим получившийся результат
```

```python
# Обернём предыдущий код в функцию, которая на вход принимает список и пороговое значение
def calc_age_avg(age_lst, thresh): # Создаём функцию
    sum_age = 0 # Вводим переменные
    cnt_age_50 = 0
    for age in age_lst: # Запускаем цикл для фильтрации условия
        if age < thresh: # Добавляем пороговое значение
            sum_age += age
            cnt_age_50 += 1
    if cnt_age_50 == 0: # Проверка на случай, если ни один элемент не удовлетворяет условию
        print(f"В списке нет элементов меньше {thresh}") # Выводим получившийся результат
        return None # Возвращаем значение

    avg_age = sum_age / cnt_age_50 # Суммируем все значения списка и делим их на длину списка
    result = round(avg_age, 2) # Округляем значение до двух знаков после запятой
    print(f"Средний возраст пользователей моложе {thresh} лет: {result}") # Выводим получившийся результат
    return result # Возвращаем значение
```

```python
calc_age_avg (age_lst, 50) # Проверяем код
```

```python
# Рассчитем среднее значение для всех элементов списка, за исключением тех, которые лежат вне интервала [Q1-1,5IQR; Q3+1,5IQR]
import numpy as np
ages = np.array(age_lst) # Преобразуем в массив NumPy для удобства вычислений
Q1 = np.percentile(ages, 25) # Находим квартили
Q3 = np.percentile(ages, 75)
IQR = Q3 - Q1 # Вычисляем IQR
lower_bound = Q1 - 1.5 * IQR # Определяем границы выбросов
upper_bound = Q3 + 1.5 * IQR
filtered_ages = ages[(ages >= lower_bound) & (ages <= upper_bound)] # Фильтруем данные — оставляем только значения в границах
avg_age = np.mean(filtered_ages) # Рассчитываем среднее и округляем до 2 знаков
result = round(avg_age, 2)
print (result)
```

```python
# Обернём предыдущий код в функцию, которая принимает на вход список и рассчитывает средний возраст, исключая из расчётов выбросы
import numpy as np

def calc_age_avg_without_outliers(age_lst): # Создаём функцию
    ages = np.array(age_lst) # Преобразуем в массив NumPy для удобства вычислений
    Q1 = np.percentile(ages, 25) # Находим квартили
    Q3 = np.percentile(ages, 75)
    IQR = Q3 - Q1 # Вычисляем IQR
    lower_bound = Q1 - 1.5 * IQR # Определяем границы выбросов
    upper_bound = Q3 + 1.5 * IQR
    filtered_ages = ages[(ages >= lower_bound) & (ages <= upper_bound)] # Фильтруем данные — оставляем только значения в границах
    if len(filtered_ages) == 0: # Проверяем, остались ли данные после фильтрации
        print("После удаления выбросов данных не осталось")
        return None
    avg_age = np.mean(filtered_ages) # Рассчитываем среднее и округляем до 2 знаков
    result = round(avg_age, 2)
    print(f"Исходные данные: {age_lst}") # Выводим дополнительную информацию для понимания процесса
    print(f"Q1 = {Q1}, Q3 = {Q3}, IQR = {IQR}")
    print(f"Границы выбросов: [{lower_bound:.2f}, {upper_bound:.2f}]")
    print(f"Данные после удаления выбросов: {filtered_ages.tolist()}")
    print(f"Средний возраст без выбросов: {result}") # Выводим получившийся результат

    return result
```

```python
calc_age_avg_without_outliers(age_lst)  # Проверяем код
```
